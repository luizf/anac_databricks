```python
%sql

CREATE SCHEMA IF NOT EXISTS anac.analytics;

CREATE EXTERNAL VOLUME IF NOT EXISTS anac.analytics.resource 
    LOCATION 'abfss://anac@lfbazure.dfs.core.windows.net/resource';
```

```python
%sql

DROP TABLE IF EXISTS anac.params.global_config;

CREATE SCHEMA IF NOT EXISTS anac.params;

CREATE TABLE IF NOT EXISTS anac.params.global_config AS
    SELECT
        'abfss://anac@lfbazure.dfs.core.windows.net/bronze/' as bronze_path,
        'abfss://anac@lfbazure.dfs.core.windows.net/silver/' as silver_path,
        'abfss://anac@lfbazure.dfs.core.windows.net/gold/' as gold_path,
        'abfss://anac@lfbazure.dfs.core.windows.net/resource/' as resource_path,
        '/Volumes/anac/analytics/resource/' as resource_path_volume,
        'dd/MM/yyyy' as date_format_br,
        'dd/MM/yyyy HH:mm:ss' as date_hour_format_br
```
