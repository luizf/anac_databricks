```python
config = spark.table("anac.params.global_config").collect()[0]

bronze_path = config["bronze_path"]
resource_path = config["resource_path"]
```

```python

df_resource = spark.read.format("json").load(resource_path + 'V_OCORRENCIA_AMPLA.json')

arquivo = 'ocorrencia_ampla'

df_resource.write \
    .mode('overwrite') \
    .format('delta') \
    .option("mergeSchema", "true") \
    .save(f"{bronze_path}/{arquivo}") 

query = f"""
    CREATE TABLE IF NOT EXISTS anac.analytics.bronze_{arquivo}
    LOCATION '{bronze_path}{arquivo}/'
"""

spark.sql(query)
```
